```python
# yield_curve_dv01_tool.py

# Import libraries and download toolkits/dependencies
# pip install pandas numpy matplotlib pandas_datareader scipy

import numpy as np
import pandas as pd
from pandas_datareader import data as pdr

# Initialize
# Treasury Constant Maturity over 1/3/6 month, 1/2/5/10/30 year indices
TENORS = {
    "DGS1MO": 1/12,
    "DGS3MO": 0.25,
    "DGS6MO": 0.50,
    "DGS1": 1.0,
    "DGS2": 2.0,
    "DGS5": 5.0,
    "DGS10": 10.0,
    "DGS30": 30.0,
}

#PURPOSE: Pulls Treasury yield curve “nodes” (yields at different maturities) from FRED and returns as-of-date and yield levels at each tenor
def get_curve_nodes(date=None, start="2018-01-01"):
    series = list(TENORS.keys())
    df = pdr.DataReader(series, "fred", start, None).dropna() #Treasury yields at different maturities. Pulls daily Treasury yields from FRED starting at start, removing missing indices
    if date is None:
        nodes = df.iloc[-1]
        asof = df.index[-1]
    else: #if date is given
        nodes = df.loc[pd.to_datetime(date)]
        asof = pd.to_datetime(date)
    # convert % -> decimal
    nodes = (nodes / 100.0).astype(float)
    return asof, nodes

def interp_rate(tenor_years, tenors, rates):
    # simple linear interpolation in tenor space
    t = np.array(tenors, dtype=float)
    r = np.array(rates, dtype=float)
    return float(np.interp(tenor_years, t, r))

def discount_factor_zero_rate(r, t, comp="cont"):
    # r is annualized zero rate (decimal)
    if comp == "cont":
        return np.exp(-r * t)
    else:
        return 1.0 / ((1.0 + r) ** t)

#price bonds off an interpolated zero curve by discounting each cash flow using continuously compounded zero rates
def bond_price_from_curve(maturity_years, coupon_rate, notional, curve_tenors, curve_rates, freq=2):
    """
    Prices a plain-vanilla bond using an interpolated zero curve.

    Builds the bond’s cash-flow schedule

    Interpolates the zero rate at each payment date

    Discounts each cash flow back to today

    Sums discounted cash flows → bond price
    """
    n_pay = int(np.round(maturity_years * freq))
    times = np.array([(i + 1) / freq for i in range(n_pay)], dtype=float)

    c = coupon_rate / freq * notional
    cashflows = np.full(n_pay, c, dtype=float)
    cashflows[-1] += notional

    dfs = []
    for t in times:
        zr = interp_rate(t, curve_tenors, curve_rates)
        dfs.append(discount_factor_zero_rate(zr, t, comp="cont"))
    dfs = np.array(dfs)

    return float(np.sum(cashflows * dfs))

#Applies a parallel shift to the entire yield curve to estimate first-order rate sensitivity (DV01, small changes)
#Adds bump_bp / 10,000 to every curve rate
#Returns the bumped curve
def curve_bump(curve_rates, bump_bp):
    return curve_rates + bump_bp / 10000.0

#Creates stylized yield-curve scenarios. Simulated steepeners and flatteners to capture non-parallel curve risk
def scenario_shift(curve_tenors, curve_rates, kind="parallel", size_bp=25):
    t = np.array(curve_tenors, dtype=float)
    r = np.array(curve_rates, dtype=float)

    if kind == "parallel":
        return r + size_bp / 10000.0

    # Define a smooth “short vs long” loading (0 at 0y -> 1 at 30y)
    load = (t - t.min()) / (t.max() - t.min())

    if kind == "steepener":
        # short down, long up
        shift = (-1 + 2 * load) * (size_bp / 10000.0)
        return r + shift

    if kind == "flattener":
        # short up, long down
        shift = (1 - 2 * load) * (size_bp / 10000.0)
        return r + shift

    raise ValueError("kind must be one of: parallel, steepener, flattener")

#Prices a portfolio of bonds off a given curve. Prices each bond using bond_price_from_curve
#and returns total portfolio value and individual bond prices
#Aggregated bond-level pricing to compute portfolio-level valuation under different curve scenarios.”
def portfolio_price(portfolio, curve_tenors, curve_rates):
    prices = []
    for bond in portfolio:
        p = bond_price_from_curve(
            maturity_years=bond["maturity_years"],
            coupon_rate=bond["coupon_rate"],
            notional=bond["notional"],
            curve_tenors=curve_tenors,
            curve_rates=curve_rates,
            freq=bond.get("freq", 2),
        )
        prices.append(p)
    return float(np.sum(prices)), np.array(prices)

#Computes portfolio DV01 (dollar value of a 1bp move). DV01 measures the portfolio’s first-order sensitivity to small parallel rate moves.
def portfolio_dv01(portfolio, curve_tenors, curve_rates, bump_bp=1.0):
    p0, _ = portfolio_price(portfolio, curve_tenors, curve_rates)
    rb = curve_bump(np.array(curve_rates), bump_bp=bump_bp)
    p1, _ = portfolio_price(portfolio, curve_tenors, rb)
    # DV01 per 1bp: price drops when rates rise => DV01 ~ (P0 - P1)
    return float(p0 - p1)

#Computes historical daily yield changes across curve tenors (FRED data). Estimated the empirical distribution of daily yield changes from historical data.
def historical_curve_changes(start="2018-01-01"):
    series = list(TENORS.keys())
    df = pdr.DataReader(series, "fred", start, None).dropna() / 100.0
    dY = df.diff().dropna()
    return dY

#Monte Carlo Interest Rate Risk Simulation
#simulated correlated yield shocks and reprice the portfolio to estimate the distribution of rate-driven PnL
def mc_rate_shocks_reprice(portfolio, base_curve_tenors, base_curve_rates, dY_hist, n_sims=20000, seed=11):
    rng = np.random.default_rng(seed)

    # Estimate shock distribution from historical daily curve changes
    mu = dY_hist.mean().values
    cov = dY_hist.cov().values

    base_p, _ = portfolio_price(portfolio, base_curve_tenors, base_curve_rates)

    shocks = rng.multivariate_normal(mu, cov, size=n_sims)  # 1-day shock
    pnl = np.zeros(n_sims)
    for i in range(n_sims):
        shocked_rates = np.array(base_curve_rates) + shocks[i]
        p1, _ = portfolio_price(portfolio, base_curve_tenors, shocked_rates)
        pnl[i] = p1 - base_p

    return {
        "base_price": base_p,
        "pnl_mean": float(np.mean(pnl)),
        "pnl_p01": float(np.quantile(pnl, 0.01)),
        "pnl_p05": float(np.quantile(pnl, 0.05)),
        "pnl_p50": float(np.quantile(pnl, 0.50)),
        "pnl_p95": float(np.quantile(pnl, 0.95)),
    }

def main():
    asof, nodes = get_curve_nodes()
    curve_tenors = np.array([TENORS[k] for k in nodes.index], dtype=float)
    curve_rates = nodes.values.astype(float)

    # current U.S. Treasuries across maturities” portfolio

    portfolio = [
        {"name": "UST 2Y", "maturity_years": 2.0, "coupon_rate": 0.045, "notional": 1_000_000, "freq": 2},
        {"name": "UST 5Y", "maturity_years": 5.0, "coupon_rate": 0.042, "notional": 1_000_000, "freq": 2},
        {"name": "UST 10Y", "maturity_years": 10.0, "coupon_rate": 0.043, "notional": 1_000_000, "freq": 2},
    ]

    p0, bond_prices = portfolio_price(portfolio, curve_tenors, curve_rates)
    dv01 = portfolio_dv01(portfolio, curve_tenors, curve_rates, bump_bp=1.0)

    print(f"As-of date: {asof.date()}")
    print(f"Portfolio price: {p0:,.2f}")
    print(f"Portfolio DV01 (per 1bp): {dv01:,.2f}")

    for scen in ["parallel", "steepener", "flattener"]:
        shocked = scenario_shift(curve_tenors, curve_rates, kind=scen, size_bp=25)
        p1, _ = portfolio_price(portfolio, curve_tenors, shocked)
        print(f"{scen.title()} +/−25bp scenario P&L: {(p1 - p0):,.2f}")

    dY_hist = historical_curve_changes(start="2018-01-01")
    mc = mc_rate_shocks_reprice(portfolio, curve_tenors, curve_rates, dY_hist, n_sims=20000)
    print("\nMonte Carlo (1-day) portfolio P&L distribution:")
    print(mc)

if __name__ == "__main__":
    main()
```
